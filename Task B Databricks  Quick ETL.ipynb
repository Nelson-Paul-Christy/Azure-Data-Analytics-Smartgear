```python
df = spark.read \
  .option("header", "true") \
  .option("inferSchema", "true") \
  .option("fs.azure.sas.raw.stnelsonsmartgear.blob.core.windows.net", "sv=2024-11-04&ss=bfqt&srt=co&sp=rwdlacuyx&se=2025-10-29T12:52:13Z&st=2025-10-28T04:37:13Z&spr=https&sig=aMtpap0p2v7cCm19n7TBMwnDh%2BL6kuYYFMcfNdkXiy0%3D") \
  .csv("wasbs://raw@stnelsonsmartgear.blob.core.windows.net/smartgear_sales.csv")
```

```python
region_summary = df.groupBy("Region").sum("Quantity") \
    .withColumnRenamed("sum(Quantity)", "TotalUnitsSold")

region_summary.show()
```

```python
region_summary.write \
  .option("fs.azure.sas.curated.stnelsonsmartgear.blob.core.windows.net", "sv=2024-11-04&ss=bfqt&srt=co&sp=rwdlacuyx&se=2025-10-29T12:52:13Z&st=2025-10-28T04:37:13Z&spr=https&sig=aMtpap0p2v7cCm19n7TBMwnDh%2BL6kuYYFMcfNdkXiy0%3D") \
  .mode("overwrite") \
  .parquet("wasbs://curated@stnelsonsmartgear.blob.core.windows.net/region_qty.parquet")
```
